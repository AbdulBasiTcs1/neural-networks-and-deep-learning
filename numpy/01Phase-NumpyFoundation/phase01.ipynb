```python
import numpy as np
from IPython.display import display, Markdown

display(Markdown("""
# NumPy Arrays

## 1. NumPy Arrays: 1D and 2D

A NumPy array is a fast, fixed-type collection of values.

- **1D array:** a single row of values
- **2D array:** rows and columns
"""))

array_1d = np.array([1, 2, 3, 4, 5])
array_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D array:")
print(array_1d)
print("Shape:", array_1d.shape)

print("\\n2D array:")
print(array_2d)
print("Shape:", array_2d.shape)

display(Markdown("""
## 2. Python Lists vs NumPy Arrays
"""))

python_list = [1, 2, 3, 4]
numpy_array = np.array([1, 2, 3, 4])

print("List multiplied by 2:", python_list * 2)
print("Array multiplied by 2:", numpy_array * 2)
print("Array addition:", numpy_array + 10)

display(Markdown("""
## 3. Creating Arrays from Scratch
"""))

zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
full = np.full((2, 2), 7)
identity = np.eye(3)
range_array = np.arange(0, 10, 2)
linespace_array = np.linspace(0, 1, 5)

print("Zeros:\\n", zeros)
print("\\nOnes:\\n", ones)
print("\\nFull:\\n", full)
print("\\nIdentity matrix:\\n", identity)
print("\\nArange:", range_array)
print("Linspace:", linespace_array)

display(Markdown("""
## 4. Vector, Matrix, and Tensor

- **Vector:** 1D array
- **Matrix:** 2D array
- **Tensor:** 3D or higher-dimensional array
"""))

vector = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])
tensor = np.arange(24).reshape(2, 3, 4)

print("Vector:\\n", vector)
print("\\nMatrix:\\n", matrix)
print("\\nTensor shape:", tensor.shape)
print(tensor)
```
